### Treating `flights_df`:


Starting off with removing  duplicates for consistent cleaning


```python
flights_df = flights_df.drop_duplicates()  #Removing the 15 duplicate records
```

Function to map airline based on flight_id to fill null values and replace "UNKNOWN" values

```python
def airline_remap(flights_df):
    """
    To fill the missing values of airline column. We are remapping airline names based on the first character of flight_id.
    All airlines follow a fixed prefix for the flight_id (i.e Flight Number) so we use it as reference
    This method overrides the airlines that are "UNKNOWN" as well as long as the flight_id is not null.
    """

    airline_mapping = {
        'A': 'Air India',
        'S': 'SpiceJet',
        '6': 'IndiGo',
        'U': 'Vistara'
    }

    flights_df['airline'] = flights_df['flight_id'].str[0].map(airline_mapping)

    return flights_df
```

```python
flights_cleaned_df = airline_remap(flights_df)
```

Verify cleaning:


```python
print('Missing values in flights_df:')
display(flights_cleaned_df.isnull().sum())
print('Duplicate rows in flights_df:')
display(flights_cleaned_df.duplicated().sum())
print('Unique values in airline:')
display(flights_cleaned_df['airline'].value_counts())
```

Verifying flight duration against departure-arrival times:

```python
def verify_duration(flights_df):
    flights_df = flights_df.copy()

    # Calculate expected duration
    calculated_duration = flights_df['arrival_time'] - flights_df['departure_time']

    # Handling overnight flights using Timedelta
    calculated_duration = calculated_duration.where(
        calculated_duration >= pd.Timedelta(0),
        calculated_duration + pd.Timedelta(days=1)
    )

    # Convert both durations to total minutes
    calculated_minutes = calculated_duration.dt.total_seconds() // 60

    duration_minutes = flights_df['duration'].apply(
        lambda x: x.hour * 60 + x.minute
    )

    # Check whether durations match at minute level
    flights_df['duration_valid'] = calculated_minutes == duration_minutes

    return flights_df
```

```python
result = verify_duration(flights_df)

anomalies = result[~result['duration_valid']]

print(anomalies[
    ['flight_id', 'departure_time', 'arrival_time', 'duration']
])
```

All flight durations are verified and are consistent with their corresponding departure and arrival times.

```python
#type normalization for all the values

# IDs / text
flights_df['flight_id'] = flights_df['flight_id'].astype('string')
flights_df['airline'] = flights_df['airline'].astype('string')
flights_df['source'] = flights_df['source'].astype('string')
flights_df['destination'] = flights_df['destination'].astype('string')

# Datetime
flights_df['departure_time'] = pd.to_datetime(
    flights_df['departure_time'],
    errors='coerce'
)

flights_df['arrival_time'] = pd.to_datetime(
    flights_df['arrival_time'],
    errors='coerce'
)

# Duration check: datetime.time / datetime.datetime -> timedelta
flights_df['duration'] = flights_df['duration'].apply(
    lambda x: pd.Timedelta(
        hours=x.hour,
        minutes=x.minute,
        seconds=x.second
    ) if pd.notna(x) else pd.NaT
)

flights_df['duration_minutes'] = (
    flights_df['duration'].dt.total_seconds() / 60
)
```

### Treating `bookings_df` and `payments_df` :
Bookings has data that can be referenced in payments to validate it. So we can clean `bookings_df` first and then we will merge it with `payments_df` to validate records on both tables.


```python
bookings_df.head()
```

Starting off by dropping PII from as it is not exactly relevant to our analysis' objective

```python
bookings_df = bookings_df.drop(
    columns=[
        'passport_number',
        'emergency_contact_name',
        'emergency_contact_phone'
    ]
)

bookings_df.head()
```

```python
payments_df.head()
```

Merging datasets:

```python
booking_payments_df = bookings_df.merge(
    payments_df[['payment_id', 'booking_id', 'amount', 'payment_method']],
    on='booking_id',
    how='left'
)
```

Mapping null values in status by referencing payment records (amount). If the value exists then it is marked as "CONFIRMED" else "INVALID"

```python

# booking_payments_df type normalization

# IDs / text
booking_payments_df['booking_id'] = booking_payments_df['booking_id'].astype('string')
booking_payments_df['passenger_id'] = booking_payments_df['passenger_id'].astype('string')
booking_payments_df['flight_id'] = booking_payments_df['flight_id'].astype('string')
booking_payments_df['status'] = booking_payments_df['status'].astype('string')
booking_payments_df['seat_number'] = booking_payments_df['seat_number'].astype('string')
booking_payments_df['payment_id'] = booking_payments_df['payment_id'].astype('string')
booking_payments_df['payment_method'] = booking_payments_df['payment_method'].astype('string')

# Datetime
booking_payments_df['booking_date'] = pd.to_datetime(
    booking_payments_df['booking_date'],
    errors='coerce'
)

# Numeric
booking_payments_df['amount'] = pd.to_numeric(
    booking_payments_df['amount'],
    errors='coerce'
)
```

```python
# Identifying missing/invalid amounts
print("Missing amounts:", payments_df['amount'].isna().sum())
```

```python
booking_payments_df['status'] = booking_payments_df['amount'].notna().map({
    True: 'CONFIRMED',
    False: 'INVALID'
})
```

```python
booking_payments_df.head()
```

```python
booking_payments_df = booking_payments_df[
    booking_payments_df['status'] != 'INVALID'
].copy()
```

```python
print('Missing values verification after merge and removal of all INVALID records:')
display(booking_payments_df.isnull().sum())
```
